# FOOD DELIVERY TIME ANALYSIS

## Visualize The Data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv("/content/Food_Delivery_Times.xls")
data.head()
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
df = data.copy()
df.shape
```

## Data Cleaning Process

```python
df.isnull().sum()
```

```python
df["Weather"] = df["Weather"].fillna("Unknown")
df["Traffic_Level"] = df["Traffic_Level"].fillna("Unknown")
df["Time_of_Day"] = df["Time_of_Day"].fillna("Unknown")
```

```python
fill_median = df["Courier_Experience_yrs"].median()
```

```python
df["Courier_Experience_yrs"] = df["Courier_Experience_yrs"].fillna(fill_median)
```

```python
df.isnull().sum()
```

```python
# Save this clean dataset into new csv file
df.to_csv("clean_food_delivery_data.csv", index=False)
```

## Exploratory Data Analysis (EDA)

```python
plt.figure(figsize=(8,6))
sns.barplot(x = "Weather", y = "Delivery_Time_min",data = df)
plt.title("Delivery Time based upon Weather")
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x = "Vehicle_Type", y = "Delivery_Time_min",data = df)
plt.title("Delivery Time by Vehicle")
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x = "Traffic_Level", y = "Delivery_Time_min",data = df)
plt.title("Delivery Time based upon Traffic Level ")
```
